```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score

# Read the dataset
df = pd.read_csv("iris.csv")

# Display first 10 rows
print(df.head(10))

# Convert dataframe to array
array = df.values

# Input features
X = array[:, 0:4]

# Target/output
y = array[:, 4]

# Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.95,
    random_state=0
)

# ---------------------------------------------------
# SVM with Linear Kernel
# ---------------------------------------------------

model = SVC(kernel='linear', random_state=0)

# Train the model
model.fit(X_train, y_train)

# Predict the test data
pred = model.predict(X_test)

print("Confusion Matrix for Linear Kernel")
conf = confusion_matrix(y_test, pred)
print(conf)

print("Accuracy for Linear Kernel")
accuracy = accuracy_score(y_test, pred)
print(accuracy)

print("Classification Report for Linear Kernel")
report = classification_report(y_test, pred)
print(report)


# ---------------------------------------------------
# SVM with RBF Kernel
# ---------------------------------------------------

model1 = SVC(kernel='rbf', random_state=0)

# Train the RBF model
model1.fit(X_train, y_train)

# Predict the test data
pred = model1.predict(X_test)

print("Confusion Matrix for RBF Kernel")
conf = confusion_matrix(y_test, pred)
print(conf)

print("Accuracy for RBF Kernel")
accuracy = accuracy_score(y_test, pred)
print(accuracy)

print("Classification Report for RBF Kernel")
report = classification_report(y_test, pred)
print(report)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score


# ==========================================================
# Read Dataset
# ==========================================================

df = pd.read_csv("iris.csv")

print("First 10 rows:")
print(df.head(10))


# ==========================================================
# Convert feature columns to numeric
# ==========================================================

df.iloc[:, 0:4] = df.iloc[:, 0:4].apply(
    pd.to_numeric,
    errors='coerce'
)

# Remove rows containing missing values
df = df.dropna()


# ==========================================================
# Input features
#
# We use only 2 features for visualization
#
# Column 0 -> Sepal Length
# Column 1 -> Sepal Width
# ==========================================================

X = df.iloc[:, 0:2].values.astype(float)


# ==========================================================
# Target
# ==========================================================

y = df.iloc[:, 4].values


# ==========================================================
# Split dataset
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=0,
    stratify=y
)


# ==========================================================
# Create Linear SVM
# ==========================================================

model = SVC(
    kernel='linear',
    C=1.0
)


# ==========================================================
# Train SVM
# ==========================================================

model.fit(X_train, y_train)


# ==========================================================
# Prediction
# ==========================================================

pred = model.predict(X_test)


# ==========================================================
# Evaluation
# ==========================================================

print("\n==========================================")
print("SVM RESULTS")
print("==========================================")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, pred))

print("\nAccuracy:")
print(accuracy_score(y_test, pred))

print("\nClassification Report:")
print(classification_report(y_test, pred))


# ==========================================================
# Create Mesh Grid
# ==========================================================

x_min = X[:, 0].min() - 0.5
x_max = X[:, 0].max() + 0.5

y_min = X[:, 1].min() - 0.5
y_max = X[:, 1].max() + 0.5


xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 500),
    np.linspace(y_min, y_max, 500)
)


# ==========================================================
# Create grid points
# ==========================================================

grid = np.c_[
    xx.ravel(),
    yy.ravel()
]


# ==========================================================
# Predict every point in the grid
# ==========================================================

Z = model.predict(grid)


# ==========================================================
# Convert class names to numbers
#
# Example:
# Iris-setosa      -> 0
# Iris-versicolor  -> 1
# Iris-virginica   -> 2
# ==========================================================

classes = model.classes_

class_to_number = {
    class_name: i
    for i, class_name in enumerate(classes)
}


Z_numeric = np.array([
    class_to_number[class_name]
    for class_name in Z
])


Z_numeric = Z_numeric.reshape(xx.shape)


# ==========================================================
# Plot decision regions
# ==========================================================

plt.figure(figsize=(10, 7))


plt.contourf(
    xx,
    yy,
    Z_numeric,
    alpha=0.2
)


# ==========================================================
# Plot the training data
# ==========================================================

for i, class_name in enumerate(classes):

    plt.scatter(
        X_train[y_train == class_name, 0],
        X_train[y_train == class_name, 1],
        label=class_name
    )


# ==========================================================
# Plot test data
# ==========================================================

plt.scatter(
    X_test[:, 0],
    X_test[:, 1],
    facecolors='none',
    edgecolors='black',
    s=80,
    label='Test Data'
)


# ==========================================================
# Plot Support Vectors
# ==========================================================

plt.scatter(
    model.support_vectors_[:, 0],
    model.support_vectors_[:, 1],
    s=150,
    facecolors='none',
    edgecolors='black',
    linewidths=2,
    label='Support Vectors'
)


# ==========================================================
# Draw SVM decision boundaries
# ==========================================================

plt.contour(
    xx,
    yy,
    Z_numeric,
    levels=[0.5, 1.5],
    linewidths=2
)


# ==========================================================
# Labels
# ==========================================================

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.title(
    "Linear SVM - Decision Boundaries and Support Vectors"
)


plt.legend()

plt.grid(True)

plt.show()
```

```python
import numpy as np

# ==========================================================
# XOR INPUT DATA
# ==========================================================

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])


# ==========================================================
# INITIAL WEIGHTS
# ==========================================================

# Input -> Hidden
w13 = -0.2
w14 = -0.4

w23 = 0.2
w24 = -0.3

# Hidden -> Output
w35 = 0.2
w45 = -0.3


# ==========================================================
# THRESHOLDS
# ==========================================================

theta3 = 0.4
theta4 = 0.1
theta5 = -0.3


# ==========================================================
# BIAS NODE
# ==========================================================

x0 = 1


# ==========================================================
# LEARNING RATE
# ==========================================================

learning_rate = 0.5


# ==========================================================
# SIGMOID FUNCTION
# ==========================================================

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


# Derivative of sigmoid
def sigmoid_derivative(x):
    return x * (1 - x)


# ==========================================================
# TRAINING
# ==========================================================

epochs = 10000

for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

        # --------------------------------------------------
        # INPUT VALUES
        # --------------------------------------------------

        x1 = X[i][0]
        x2 = X[i][1]

        target = Y[i][0]


        # ==================================================
        # FORWARD PROPAGATION
        # ==================================================

        # Hidden neuron x3
        net3 = (x1 * w13) + (x2 * w23) - theta3
        x3 = sigmoid(net3)


        # Hidden neuron x4
        net4 = (x1 * w14) + (x2 * w24) - theta4
        x4 = sigmoid(net4)


        # Output neuron x5
        net5 = (x3 * w35) + (x4 * w45) - theta5
        x5 = sigmoid(net5)


        # ==================================================
        # ERROR
        # ==================================================

        error = target - x5

        total_error += error ** 2


        # ==================================================
        # BACKPROPAGATION
        # ==================================================

        # Output layer delta
        delta5 = error * sigmoid_derivative(x5)


        # Hidden layer deltas
        delta3 = sigmoid_derivative(x3) * (delta5 * w35)

        delta4 = sigmoid_derivative(x4) * (delta5 * w45)


        # ==================================================
        # UPDATE HIDDEN -> OUTPUT WEIGHTS
        # ==================================================

        w35 = w35 + learning_rate * delta5 * x3

        w45 = w45 + learning_rate * delta5 * x4


        # ==================================================
        # UPDATE OUTPUT THRESHOLD
        # ==================================================

        theta5 = theta5 - learning_rate * delta5


        # ==================================================
        # UPDATE INPUT -> HIDDEN WEIGHTS
        # ==================================================

        w13 = w13 + learning_rate * delta3 * x1

        w23 = w23 + learning_rate * delta3 * x2

        w14 = w14 + learning_rate * delta4 * x1

        w24 = w24 + learning_rate * delta4 * x2


        # ==================================================
        # UPDATE HIDDEN THRESHOLDS
        # ==================================================

        theta3 = theta3 - learning_rate * delta3

        theta4 = theta4 - learning_rate * delta4


    # Print error every 1000 epochs
    if epoch % 1000 == 0:
        print("Epoch:", epoch, "Error:", total_error)


# ==========================================================
# FINAL WEIGHTS
# ==========================================================

print("\nFinal Weights:")
print("w13 =", w13)
print("w14 =", w14)
print("w23 =", w23)
print("w24 =", w24)
print("w35 =", w35)
print("w45 =", w45)

print("\nFinal Thresholds:")
print("theta3 =", theta3)
print("theta4 =", theta4)
print("theta5 =", theta5)


# ==========================================================
# TEST XOR
# ==========================================================

print("\nXOR OUTPUT:")

for i in range(len(X)):

    x1 = X[i][0]
    x2 = X[i][1]

    # Forward propagation

    net3 = (x1 * w13) + (x2 * w23) - theta3
    x3 = sigmoid(net3)

    net4 = (x1 * w14) + (x2 * w24) - theta4
    x4 = sigmoid(net4)

    net5 = (x3 * w35) + (x4 * w45) - theta5
    x5 = sigmoid(net5)

    # Convert probability to 0 or 1
    prediction = 1 if x5 >= 0.5 else 0

    print(
        "Input:", x1, x2,
        "Hidden:", round(x3, 4), round(x4, 4),
        "Output:", round(x5, 4),
        "Prediction:", prediction,
        "Target:", Y[i][0]
    )
```
