```python

class Student:
    def __init__(self, regno, name):
        self.regno = regno
        self.name = name

    def display(self):
        print("Register Number:", self.regno)
        print("Name:", self.name)

s1 = Student(227616, "Sam")

print(s1.regno)
print(s1.name)
s1.display()
```
