```python
import pandas as pd
import numpy as np


data = {
    "CGPA": [">=9", ">=9", ">=8", ">=9"],
    "Interactive": ["Yes", "Yes", "No", "Yes"],
    "Practical_Knowledge": ["Excellent", "Good", "Good", "Good"],
    "Communication_Skills": ["Good", "Good", "Good", "Good"],
    "Logical_Thinking": ["Fast", "Fast", "Fast", "Slow"],
    "Interest": ["Yes", "Yes", "No", "No"],
    "Job_Offer": ["Yes", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

print("\n================ TRAINING DATASET ================\n")
print(df)

concepts = np.array(df.iloc[:, :-1])
target = np.array(df.iloc[:, -1])


S = concepts[0].copy()

G = [["?" for i in range(len(S))] for j in range(len(S))]

print("\nInitial Specific Hypothesis (S)")
print(S)

print("\nInitial General Hypothesis (G)")
for row in G:
    print(row)

for i, h in enumerate(concepts):

    print("\n===================================================")
    print("Training Instance", i + 1)
    print(h)
    print("Class Label :", target[i])

    if target[i] == "Yes":

        print("\nPositive Example")
        print("Generalizing Specific Hypothesis...")

        for x in range(len(S)):
            if h[x] != S[x]:
                S[x] = "?"
                G[x][x] = "?"

        print("\nUpdated Specific Hypothesis (S)")
        print(S)

        print("\nUpdated General Hypothesis (G)")
        for row in G:
            print(row)

    else:

        print("\nNegative Example")
        print("Specializing General Hypothesis...")

        for x in range(len(S)):
            if h[x] != S[x]:
                G[x][x] = S[x]
            else:
                G[x][x] = "?"

        print("\nSpecific Hypothesis (S)")
        print(S)

        print("\nUpdated General Hypothesis (G)")
        for row in G:
            print(row)


G_final = []

for row in G:
    if row != ["?"] * len(S):
        G_final.append(row)


print("\n===================================================")
print("FINAL SPECIFIC HYPOTHESIS")
print(S)

print("\nFINAL GENERAL HYPOTHESIS")

if len(G_final) == 0:
    print("No specific general hypothesis remains.")
else:
    for row in G_final:
        print(row)
```
