```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    "week": [1, 2, 3, 4, 5],
    "sales": [1.2, 1.8, 2.6, 3.2, 3.8]
}

# Create DataFrame
df = pd.DataFrame(data)

# Independent variable
X = df[["week"]]

# Dependent variable
y = df["sales"]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict values
predictions = model.predict(X)

# Print results
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"Sales = {model.coef_[0]:.2f} × Week + {model.intercept_:.2f}")

print(f"R² Score: {model.score(X, y):.4f}")

# Plot graph
plt.figure(figsize=(8,5))
plt.scatter(X["week"], y, color="green", label="Actual Data")
plt.plot(X["week"], predictions, color="red", linewidth=2, label="Regression Line")
plt.title("Linear Regression: Week vs Sales")
plt.xlabel("Week")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression

# Create a random dataset
X, y = make_regression(
    n_samples=50,
    n_features=1,
    noise=10,
    random_state=43
)

# Display the data points
plt.figure(figsize=(6, 4))
plt.scatter(X, y, color="green", label="Data Points")

# Create the Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict the values
predictions = model.predict(X)

# Print results
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"y = {model.coef_[0]:.2f}x + {model.intercept_:.2f}")

print(f"\nR² Score: {model.score(X, y):.4f}")

# Plot regression line
plt.plot(X, predictions, color="red", linewidth=2, label="Regression Line")

# Add title and labels
plt.title("Linear Regression using Random Dataset")
plt.xlabel("X")
plt.ylabel("y")

# Show legend and grid
plt.legend()
plt.grid(True)

# Display graph
plt.show()
```

```python
# Import libraries
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

print("Multiple Linear Regression")

# Create a random dataset
X, y = make_regression(
    n_samples=50,
    n_features=2,
    noise=0.1,
    random_state=43
)

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Print intercept
print("Intercept:", model.intercept_)

# Print coefficients
print("\nCoefficients:")
for i, coef in enumerate(model.coef_, start=1):
    print(f"X{i}: {coef:.4f}")

# Print regression equation
print("\nRegression Equation:")
print(
    f"y = {model.coef_[0]:.4f}X1 + "
    f"{model.coef_[1]:.4f}X2 + "
    f"{model.intercept_:.4f}"
)

# Print R² Score
print(f"\nR² Score: {model.score(X, y):.4f}")
```

```python
import numpy as np
from sklearn.decomposition import PCA

# Create a 2x2 matrix
X = np.array([[2, 6],
              [1, 7]])

print("Original Matrix:")
print(X)
print("Shape:", X.shape)

# Create PCA object
pca = PCA(n_components=2)

# Transform the data
X_pca = pca.fit_transform(X)

print("\nTransformed Matrix:")
print(X_pca)
print("Shape:", X_pca.shape)

# Reconstruct the original data
X_reconstructed = pca.inverse_transform(X_pca)

print("\nReconstructed Matrix (Inverse Transform):")
print(X_reconstructed)

# Explained Variance Ratio
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Singular Values
print("\nSingular Values:")
print(pca.singular_values_)

# Principal Components
print("\nPrincipal Components:")
print(pca.components_)

# Mean of Dataset
print("\nMean of Dataset:")
print(pca.mean_)
```

```python
import numpy as np

# Create a matrix
A = np.array([[1, 2],
              [3, 4],
              [5, 6]])

print("Original Matrix (A):")
print(A)

# Perform Singular Value Decomposition (SVD)
U, S, VT = np.linalg.svd(A)

print("\nMatrix U:")
print(U)

print("\nSingular Values (S):")
print(S)

print("\nMatrix V^T:")
print(VT)

# Create Sigma matrix
Sigma = np.zeros((A.shape[0], A.shape[1]))
np.fill_diagonal(Sigma, S)

print("\nSigma Matrix:")
print(Sigma)

# Reconstruct the original matrix
A_reconstructed = U @ Sigma @ VT

print("\nReconstructed Matrix:")
print(A_reconstructed)

# Verify reconstruction
print("\nVerification:", np.allclose(A, A_reconstructed))
```
